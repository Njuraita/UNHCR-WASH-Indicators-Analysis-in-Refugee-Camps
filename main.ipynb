Dataset Columns Overview

The dataset contains Water, Sanitation and Hygiene (WASH) indicators collected across refugee camps and settlements in different countries. It includes information on reporting periods, geographical locations, refugee populations, water availability, water quality, sanitation facilities and hygiene services.

The dataset will be explored to understand its structure, variables, data types, completeness and suitability for monitoring and evaluating WASH service delivery.

2.2 Description of Dataset Variable

A. Identification and Reporting Information

| Variable | Description |
|---|---|
| `form_id` | Unique identifier for a submitted data collection form or record. |
| `start_date` | Start date of the reporting period or data collection. |
| `end_date` | End date of the reporting period or data collection. |
| `location_id` | Unique identification code assigned to a reporting location. |
| `location_name` | Name of the refugee camp, settlement or reporting location. |
| `country` | Country where the reporting location is situated. |
| `post_emergency` | Indicates whether the location or response is in the post-emergency phase. |
| `reporting_monthly` | Indicator relating to monthly reporting. |
| `reporting_annual` | Indicator relating to annual reporting. |

B. Water Availability and Access

| Variable | Description |
|---|---|
| `persons_per_handpump` | Average number of people served by one hand pump. |
| `persons_per_tap` | Average number of people served by one water tap. |
| `liters_per_person_per_day` | Average quantity of water available per person per day, measured in litres. |
| `liters_per_person_household` | Household-level indicator of water quantity available per person per day. |
| `potable_water_storage_10l` | Indicator relating to household access to or availability of potable-water storage containers of at least 10 litres. |
| `protected_water_sources` | Indicator measuring access to protected water sources designed to reduce contamination. |

C. Water Quality

| Variable | Description |
|---|---|
| `non_chlorinated_0_cfu` | Indicator relating to non-chlorinated water samples with zero colony-forming units (CFU) under the specified testing conditions. |
| `chlorinated_safe_water_quality` | Indicator measuring whether chlorinated water meets applicable safe-water quality requirements. |

D. Sanitation and Hygiene Facilities

| Variable | Description |
|---|---|
| `households_with_toilet` | Proportion or percentage of households with access to a toilet. |
| `persons_per_toilet` | Average number of people sharing one toilet. |
| `persons_per_shower` | Average number of people sharing one shower facility. |
| `persons_per_hygiene_promoter` | Average number of people served by one hygiene promoter. |
| `household_toilet` | Indicator relating to toilet access or availability at household level. |
| `defecate_in_toilet` | Indicator measuring the use of toilets for defecation rather than open defecation. |
| `access_to_soap` | Indicator measuring household access to soap for handwashing and personal hygiene. |
| `solid_waste_disposal_access` | Indicator measuring access to appropriate solid-waste disposal services or facilities. |

E. Population Information

| Variable | Description |
|---|---|
| `refugee_pop` | Refugee population recorded for the reporting location. |

F. Menstrual Hygiene

| Variable | Description |
|---|---|
| `menstrual_hygiene_satisfaction` | Indicator measuring satisfaction with facilities and services available for managing menstrual hygiene and menstrual waste. |

Preliminary Understanding of Key Indicators

The variables can be grouped into four main Monitoring and Evaluation (M&E) areas:

1. **Water availability:** Water quantity per person and access to water points.
2. **Water quality:** Safety of drinking water.
3. **Sanitation:** Toilet availability, toilet sharing and toilet usage.
4. **Hygiene:** Access to soap, menstrual hygiene and solid-waste disposal.

```python

def country_to_continent(country):
    try:
        country_alpha2 = pc.country_name_to_country_alpha2(country)
        continent_code = pc.country_alpha2_to_continent_code(country_alpha2)
        continent_name = pc.convert_continent_code_to_continent_name(continent_code)
        return continent_name
    except:
        return "Unknown"

df["continent"] = df["country"].apply(country_to_continent)
```

```python
df["continent"].unique()
```

```python
# Filter rows where continent is "Unknown"
unknown_countries = df[df["continent"] == "Unknown"][["country", "location_name"]]
print("Unknown countries:", unknown_countries["country"].unique())
```

```python
# Group by country and list unique camps
countries_camps = df.groupby('country')['location_name'].unique()

# Display the result
print(countries_camps)
```

```python
# Group by country and get unique camps
countries_camps = df.groupby('country')['location_name'].unique()

# Count the number of camps per country
num_camps_per_country = countries_camps.apply(len)

# Display result
print(num_camps_per_country)
```
